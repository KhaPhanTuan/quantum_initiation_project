```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)
```

```python
print(qc.draw())
```

```python
sim = AerSimulator()
compiled = transpile(qc, sim)
counts = sim.run(compiled, shots=1024).result().get_counts()
print(counts)
plot_histogram(counts)
```
